```python
Combining the data of all these 3 dfs (Union + distinct):
```

```python
bad_data_df = rep_cust_data.select("mem_id") \
.union(rep_cust_delinq_data_df.select("mem_id")) \
.union(rep_cust_defaulters_data_df.select("mem_id"))
```

```python
bad_cust_data_final_df = bad_data_df.distinct()
```

```python
bad_cust_data_final_df.count()
```

```python
bad_cust_data_final_df.repartition(1).write \
.format("parquet") \
.mode("overwrite") \
.option("header", True) \
.option("path","/user/itv024771/landing_club/bad/bad_cust_data_final") \
.save()
```

```python
bad_cust_data_final_df.createOrReplaceTempView("loan_bad_data")
```

```python
Now removing the bad data union from our external tables: 
```

```python
cust_df = spark.sql("select * from itv024771_lending_club.customers where mem_id NOT IN (select mem_id from loan_bad_data)")
```

```python
loan_delinq_df = spark.sql("select * from itv024771_lending_club.loan_defaul_del where mem_id NOT IN (select mem_id from loan_bad_data)")
```

```python
loan_defaulters_df = spark.sql("select * from itv024771_lending_club.loan_defaulters where mem_id NOT IN (select mem_id from loan_bad_data)")
```

```python
cust_df.write \
.format("parquet") \
.mode("overwrite") \
.option("path","/user/itv024771/landing_club/cleaned/final/customer_parquet") \
.save()
```

```python
loan_delinq_df.write \
.format("parquet") \
.mode("overwrite") \
.option("path","/user/itv024771/landing_club/cleaned/final/loan_delinq_parquet") \
.save()
```

```python
loan_defaulters_df.write \
.format("parquet") \
.mode("overwrite") \
.option("path","/user/itv024771/landing_club/cleaned/final/loan_defaulters_parquet") \
.save()
```
