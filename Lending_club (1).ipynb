```python
cust_emp_updated_df = dis_cust_df.withColumn("emp_length",regexp_replace(col("emp_length"),"(\D)",""))
```

```python
cust_emp_updated_df.show()
```

```python
cust_emp_updated_df.printSchema()
```

```python
cust_emp_dtype_df = cust_emp_updated_df.withColumn("emp_length",cust_emp_updated_df.emp_length.cast('int'))
```

```python
cust_emp_dtype_df.printSchema()
```

```python
cust_emp_dtype_df.show()
```

```python
cust_emp_dtype_df.filter("emp_length is null").count()
```

```python
cust_emp_dtype_df.createOrReplaceTempView("customer")
```

```python
avg_emp_length = spark.sql("select floor(avg(emp_length)) as avg_emp from customer").collect()
```

```python
print(avg_emp_length)
```

```python
avg_emp = avg_emp_length[0][0]
```

```python
cust_emplength_new= cust_emp_dtype_df.na.fill(avg_emp,subset=['emp_length'])
```

```python
cust_emplength_new.show()
```

```python
cust_emplength_new.filter("emp_length is null").count()
```
